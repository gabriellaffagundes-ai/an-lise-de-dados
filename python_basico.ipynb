#Tipos de dados

Strings (str): texto

Number (int): número inteiro

Float: número real com fração

Complex: número complexo. Exemplo: x + yj = z

Bool: verdadeiro ou falso

#Por que especificar o tipo de variável?

```python

#O tipo da variável deve ser especificado para não gerar erros

m = float(input('digite um número: '))
n = float(input('digite o dia do seu aniversário: '))
print('a soma vale', m + n)

#Exemplo em que a falta de especificação acarreta erro de cálculo

p = input('horário de nascimento: ')
q = input('seu número favorito é: ')
print('A soma vale', p + q)



```

```python
#Criar função

def caixa_preta():
  print('teste')
  print('teste2')

#Executar função

caixa_preta()

#Função com argumento

def média(a,b):
  print((a+b)/2)

média(10, 3)

#Função Return
##Definir os valores antes de criar a função

x = 15
y = 1
def soma(x, y):
  return x + y

c = soma(x, y)
print(c)

#Função


valores = [2, 4, 6, 8]
def exponencial(valores):
  exponenciais = []

#For_in é usado para aplicar a operação sobre todos os elementos fornecidos

  for m in valores:
    exponenciais.append(m**2)
  return exponenciais

resultado = exponencial(valores)

for  n in resultado:
  print(n)
```
