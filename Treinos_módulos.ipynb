```python
#Conversão de real para dolar

## 1 real = 0.18 dolar

r = float(input('valor em reais: '))
d = r * 0.18
print('valor em dolar =', d)
```

```python
#Cálculo da quantidade de tinta para pintar um muro
##Cada litro de tinta pinta 3m^2

h = float(input('qual a altura do muro? '))
c = float(input('qual a largura do muro? '))
a = h * c
t = a/3
print('a quantidade de tinta necessária é {}L'.format(t))
```

```python
#Mostre o preço de um produto com 20% de desconto
##p = preço sem desconto
##pd = preço com desconto

p = float(input('qual o preço do produto? '))
pd = p * 0.8
print('com desconto, o consumidor deve pagar R${}'.format(pd))
```
