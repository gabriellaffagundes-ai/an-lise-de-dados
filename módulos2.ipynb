```python
#import math
math.sqrt(16)

import math
math.radians(2*(math.pi))

#Insira número real, receba inteiro

x = float(input('digite um número real: '))
print('a parte inteira desse número é', math.trunc(x))
```

```python
#Calculando a hipotenusa

a = float(input('medida do cateto adjacente '))
b = float(input('medida do cateto oposto '))
c = math.hypot(a, b)
print('a hipotenusa mede', c)
```

```python
#Seno, cosseno e tangente de um ângulo
##O ângulo é padronizado como radiano
x = float(input('digite o ângulo em graus: '))
sin = math.sin(math.radians(x))
cos = math.cos(math.radians(x))
tan = math.tan(math.radians(x))
print('para x, temos: sen(x) = {:.2f}, cos(x) = {:.2f}, tan(x) = {:.2f}'.format(sin, cos, tan))
```

```python
from math import sin, cos, tan, radians

x = float(input('digite o ângulo em graus: '))
sin = sin(radians(x))
cos = cos(radians(x))
tan = tan(radians(x))

print('o seno, cosseno e tangente de x são: {}, {} e {}'.format(sin, cos, tan))

```

```python
#Lista
#Sorteio
import random
a = str(input('insira o nome do concorrente: '))
b = str(input('insira o nome do concorrente: '))
c = str(input('insira o nome do concorrente: '))
d = str(input('insira o nome do concorrente: '))

lista = [a, b, c, d]

sorteado = random.choice(lista)

print('o escolhido foi:', sorteado)
```

```python
#Sorteio de ordem de apresentação

g = input('insira o nome: ')
h = input('insira o nome: ')
i = input('insira o nome: ')
j = input('insira o nome: ')

lista = [g, h, i, j]

from random import choice

s1 = choice(lista)
s2 = choice(lista)
s3 = choice(lista)
s4 = choice(lista)

print('a ordem de apresentação será: {}, {}, {} e {}'.format(s1, s2, s3, s4))

#Esse modo pode apresentar erros, pois pode repetir um mesmo nome
```
