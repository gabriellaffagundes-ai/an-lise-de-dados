#Manipulando cadeia de caracteres

Frase - cada caractere tem um número a ele associado, de 0 a n.
Ex.: Aprendendo python
     0123............17

#Fatiamento

frase[12] identifica a letra y

frase[3:8] associa o caractere 3 até o 7

frase[3:8:2] II   excluindo um caractere do 3 ao 7

frase[0:5] = frase[:5]

frase[7:] = frase[7:18]

frase[9::3] = associa a partir do caractere 9 até o último, excluindo de 2 em 2 algarismos

#Análise

len(frase) indica a qtd de caracteres

frase.count('o') - conta qtd de 'o'

frase.count('o', 0, 13) - conta qtd de 'o' do caractere 0 ao 12

frase.find('den') - indica a posição onde começam esses caracteres

frase.find('Gabriella") - retorna -1, já que não existe esse conjunto de letras na frase em questão

'Aprendendo' in frase - retorna True, pois tem isso na frase

#Transformação

frase.replace('Aprendendo', 'Compreendendo') - troca uma palavra pela outra

frase.upper() - transforma todas as letras para maiúsculas

frase.lower() - II                              minúsculas

frase.capitalize() - deixa apenas a 1a letra maiúscula

frase.title() - deixa a 1a letra de cada palavra em maiúsculo

frase.strip() - remove os espaços desnecessários

frase.rstripe() - II                            à direita

frase.lstripe() - II                              esquerda

#Divisão

frase.split()

separa as palavras e começa a numeração do 1o caractere de cada palavra da frase com 0

Aprendendo Python
012............16

Aprendendo     Python
01.......9     0....5
**  0             1**

#Junção

' '.join(frase)
faz o oposto da divisão

```python
frase = 'Os buracos negros ainda são um mistério para a Astronomia!'.upper()

x = frase.count('A')
print("o caractere 'A' aparece quantas vezes? ", x)

y = frase.find('A', 0, )
print("a letra 'a' aparece pela primeira vez na posição ", y)

z = frase.rfind('A', 0, )
print("a letra 'a' aparece pela última vez na posição ", z)
```
