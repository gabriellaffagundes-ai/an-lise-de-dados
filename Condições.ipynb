#Import sleep
Inclui um tempo de espera para que a informação seja exibida na tela

```python
from random import randint
from time import sleep

comp = randint(0, 10)
print('_' * 20)
print('Adivinhe em qual número estou pensando')
print('_' * 20)

humano = int(input('Acho que o computador está pensando no número: '))

print('PROCESSANDO...')
sleep(3)

if comp == humano:
  print('UAU! Você é muito bom nisso!!!')
else:
  print('Tenta a sorte!')
```

```python
#Multa radar

velocidade = float(input('velocidade registrada no radar: '))
print('Sua velocidade foi {} km/h'.format(velocidade))

excedido = velocidade - 80

multa = 7 * excedido

if velocidade > 90.0:
  print('Você será multado!')
  print('O valor da multa será {:.2f} reais'.format(multa))
else:
  print('Parabéns por respeitar as leis de trânsito!')

```

```python
#Paridade
número = int(input('digite um número: '))

if número % 2 == 0:
  print('O número {} é par'.format(número))
else:
  print('O número {} é ímpar'.format(número))
```

```python
d = float(input('qual a distância, em km, da sua viagem? '))

if d > 200:
  t = 0.45 * d
  print('O valor da passagem é R${:.3f}'.format(t))
else:
  t = 0.5 * d
  print('O valor da passagem é R${:.3f}'.format(t))
```

```python
#Comparação entre números

a = int(input('digite um número: '))
b = int(input('digite um número: '))
c = int(input('digite um número: '))

if a > b and a > c:
  print('O {} é o maior número'.format(a))

if b > c and b > a:
  print('O {} é o maior número'.format(b))

if c > b and c > a:
  print('O {} é o maior número'.format(c))

if a < b and a < c:
    print('O {} é o menor número'.format(a))

if b < c and b < a:
  print('O {} é o menor número'.format(b))

if c < a and c < b:
  print('O {} é o menor número'.format(c))
```

```python
#Desigualdade triangular
m = float(input('digite o valor de um dos lados do triângulo: '))
print('_=_'*30)
n = float(input('digite o valor de um dos lados do triângulo: '))
print('_=_'*30)
o = float(input('digite o valor de um dos lados do triângulo: '))
print('_=_'*30)

if m < (n + o) and n < (m + o) and o < (m + n):
  print('É possível formar um triângulo')
else:
  print('Esse conjunto de medidas não obedece a desigualdade triangular')
```
